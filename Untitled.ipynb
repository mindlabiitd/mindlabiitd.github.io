```python
import bibtexparser, yaml

with open("_publications/Publications.txt") as bibtex_file:
    bib_db = bibtexparser.load(bibtex_file)

# Convert the 'year' field to int for proper numeric sort.
# Sort descending (recent → oldest).
entries_sorted = sorted(
    bib_db.entries,
    key=lambda e: int(e.get("year", 0)),
    reverse=True
)

with open("_publications/publications.yml", "w") as out:
    yaml.dump(bib_db.entries, out, sort_keys=False)
```
