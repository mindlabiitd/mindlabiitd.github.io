```python
import yaml

# Load the YAML
with open("_data/publications.yml") as f:
    pubs = yaml.safe_load(f)

def abbreviate_author(author_str):
    authors = [a.strip() for a in author_str.split(" and ")]
    abbreviated = []

    for a in authors:
        if "," in a:
            last, first = a.split(",", 1)
            last = last.strip()
            first = first.strip()
            # split first names into words, take first letter of each, join with ". "
            initials = ". ".join([name[0] for name in first.split() if name]) + "."
            abbreviated.append(f"{initials} {last}")
        else:
            abbreviated.append(a)  # fallback if not in "Lastname, Firstname" format

    # Join all authors with commas
    return ", ".join(abbreviated)

# Update each publication
for pub in pubs:
    if "author" in pub:
        pub["author_abbrev"] = abbreviate_author(pub["author"])

# Write out updated YAML (keeps original too)
with open("_data/publicationsabbrev.yml", "w") as f:
    yaml.dump(pubs, f, sort_keys=False)
```
